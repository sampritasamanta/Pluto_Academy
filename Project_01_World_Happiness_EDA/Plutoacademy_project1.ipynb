```python
df2015_clean = df2015.rename(columns={
    'Happiness Rank': 'Happiness Rank',
    'Happiness Score': 'Happiness Score',
    'Economy (GDP per Capita)': 'GDP per Capita',
    'Family': 'Social Support',
    'Health (Life Expectancy)': 'Health',
    'Freedom': 'Freedom',
    'Generosity': 'Generosity',
    'Trust (Government Corruption)': 'Corruption'
})

df2015_clean['Year'] = 2015
```

```python
df2016_clean = df2016.rename(columns={
    'Happiness Rank': 'Happiness Rank',
    'Happiness Score': 'Happiness Score',
    'Economy (GDP per Capita)': 'GDP per Capita',
    'Family': 'Social Support',
    'Health (Life Expectancy)': 'Health',
    'Freedom': 'Freedom',
    'Generosity': 'Generosity',
    'Trust (Government Corruption)': 'Corruption'
})

df2016_clean['Year'] = 2016
```

```python
df2017_clean = df2017.rename(columns={
    'Happiness.Rank': 'Happiness Rank',
    'Happiness.Score': 'Happiness Score',
    'Economy..GDP.per.Capita.': 'GDP per Capita',
    'Family': 'Social Support',
    'Health..Life.Expectancy.': 'Health',
    'Freedom': 'Freedom',
    'Generosity': 'Generosity',
    'Trust..Government.Corruption.': 'Corruption'
})

df2017_clean['Year'] = 2017
```

```python
df2018_clean = df2018.rename(columns={
    'Overall rank': 'Happiness Rank',
    'Country or region': 'Country',
    'Score': 'Happiness Score',
    'GDP per capita': 'GDP per Capita',
    'Social support': 'Social Support',
    'Healthy life expectancy': 'Health',
    'Freedom to make life choices': 'Freedom',
    'Generosity': 'Generosity',
    'Perceptions of corruption': 'Corruption'
})

df2018_clean['Year'] = 2018
```

```python
df2019_clean = df2019.rename(columns={
    'Overall rank': 'Happiness Rank',
    'Country or region': 'Country',
    'Score': 'Happiness Score',
    'GDP per capita': 'GDP per Capita',
    'Social support': 'Social Support',
    'Healthy life expectancy': 'Health',
    'Freedom to make life choices': 'Freedom',
    'Generosity': 'Generosity',
    'Perceptions of corruption': 'Corruption'
})

df2019_clean['Year'] = 2019
```

```python
common_columns = [
    'Year',
    'Country',
    'Happiness Rank',
    'Happiness Score',
    'GDP per Capita',
    'Social Support',
    'Health',
    'Freedom',
    'Generosity',
    'Corruption'
]
```

```python
df_happiness = pd.concat([
    df2015_clean[common_columns],
    df2016_clean[common_columns],
    df2017_clean[common_columns],
    df2018_clean[common_columns],
    df2019_clean[common_columns]
], ignore_index=True)
```

## Data Cleaning Decisions

1. The five yearly CSV files (2015–2019) had different column names, so the columns were renamed to a common structure before combining the datasets.

2. Only the relevant variables were retained: Year, Country, Happiness Rank, Happiness Score, GDP per Capita, Social Support, Health, Freedom, Generosity, and Corruption.

3. The single missing value in the 2018 dataset was handled during data preparation, resulting in no missing values in the final combined dataset.

4. Data types were checked to ensure numerical variables were stored as numeric values and Country was stored as text.

5. Duplicate rows were checked and no duplicates were found.

6. After cleaning, the final dataset contained 782 rows and 10 columns, with 0 missing values and 0 duplicate rows.

### Q1. How did the average Happiness Score change from 2015 to 2019?

The average Happiness Score remained relatively stable between 2015 and 2019.
It reached its lowest value in 2017 (5.354) and its highest value in 2019 (5.407).
Overall, the average score increased slightly from 2015 to 2019.

### Q4. Is Social Support associated with Happiness Score?

Social Support has a moderate-to-strong positive correlation with Happiness Score, with a correlation coefficient of 0.6488. This indicates that countries with higher Social Support generally tend to have higher Happiness Scores.


```python
correlation = df_happiness[
    ['Happiness Score', 'GDP per Capita', 'Social Support',
     'Health', 'Freedom', 'Generosity', 'Corruption']
].corr()['Happiness Score'].sort_values(ascending=False)

correlation
```

```python
correlation_factors = correlation.drop('Happiness Score')

plt.figure(figsize=(9, 5))

plt.bar(
    correlation_factors.index,
    correlation_factors.values
)

plt.title('Correlation of Factors with Happiness Score')
plt.xlabel('Factor')
plt.ylabel('Correlation Coefficient')
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

### Q5. Which factors are most strongly correlated with Happiness Score?

GDP per Capita has the strongest positive correlation with Happiness Score (0.7893), followed by Health (0.7425) and Social Support (0.6488). Freedom also shows a positive correlation (0.5513), while Corruption has a weaker positive correlation (0.3973). Generosity has the weakest correlation among the factors examined (0.1376).

## Key Insights

1. **Happiness remained relatively stable from 2015 to 2019.**
   The average Happiness Score ranged from 5.354 in 2017 to 5.407 in 2019, showing only a small overall change.

2. **Finland ranked as the happiest country in 2019.**
   Finland had the highest Happiness Score in 2019, with a score of 7.769, followed by Denmark and Norway.

3. **GDP per Capita has the strongest association with Happiness Score.**
   GDP per Capita had the highest correlation with Happiness Score among the examined factors, with a correlation coefficient of 0.7893.

4. **Health and Social Support are also strongly associated with happiness.**
   Health had a correlation of 0.7425, while Social Support had a correlation of 0.6488 with Happiness Score.

5. **Generosity has the weakest association among the examined factors.**
   Generosity had a correlation of only 0.1376 with Happiness Score, considerably lower than GDP per Capita, Health, Social Support, and Freedom.

## Finding That Surprised Me the Most

The finding that surprised me the most was the relatively strong correlation between GDP per Capita and Happiness Score (0.7893). I expected social factors such as Social Support to have a stronger relationship with happiness. However, GDP per Capita showed the strongest association among the factors analyzed. This suggests that economic conditions are an important factor associated with national happiness, although correlation does not imply causation.

## Conclusion

The World Happiness dataset was analyzed for the years 2015–2019. The analysis showed that Happiness Scores remained relatively stable over the five-year period. Finland had the highest Happiness Score in 2019.

The correlation analysis showed that GDP per Capita had the strongest positive association with Happiness Score, followed by Health and Social Support. Freedom also showed a positive association, while Generosity had the weakest association among the factors examined.

Overall, the analysis suggests that economic conditions, health, and social support are important factors associated with national Happiness Scores. However, correlation does not imply causation, so these results should be interpreted as associations rather than causal relationships.
